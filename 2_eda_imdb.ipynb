### 21. Obten los valores únicos de la variable Rating y la frecuencia total de cada uno de estos valores. Posteriormente crea un nuevo dataframe con essos valores

```python
frecuencia_ratings = df.Rating.value_counts()
df_frecuencias = frecuencia_ratings.reset_index()
df_frecuencias.columns = ['Rating', 'frecuencia_total']
df_frecuencias = df_frecuencias.sort_values(by='Rating').reset_index(drop=True)
df_frecuencias
```

### 22. Ordena el nuevo dataframe por la variable `rating`

```python
df_frecuencias = df_frecuencias.sort_values(by='Rating')
df_frecuencias = df_frecuencias.reset_index(drop=True)
df_frecuencias
```

### 23. Crea una gráfica de barras con este nuevo dataframe ordenado

```python
import seaborn as sns
plt.figure(figsize=(14, 6))
sns.barplot(x='Rating', y='frecuencia_total', data=df_frecuencias, color='royalblue')
plt.xticks(rotation=45, horizontalalignment='right')
plt.title('Frecuencia Total de Películas por Calificación (Rating)', fontsize=16, fontweight='bold')
plt.xlabel('Calificación (Rating)', fontsize=12)
plt.ylabel('Cantidad de Películas', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### 24. Crea la matriz de correlación del dataframe de películas

```python
df_numerico = df.select_dtypes(include=['float64', 'int64'])
matriz_corr = df_numerico.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(matriz_corr, 
            annot=True, 
            cmap='coolwarm', 
            fmt=".2f", 
            linewidths=0.5, 
            vmin=-1, 
            vmax=1)
plt.title('Matriz de Correlación del DataFrame de Películas', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 25. Grafica la matriz de correlación utilizando `matshow()`

```python
df_numerico = df.select_dtypes(include=['float64', 'int64'])
matriz_corr = df_numerico.corr()

fig, ax = plt.subplots(figsize=(10, 8))
cax = ax.matshow(matriz_corr, cmap='coolwarm', vmin=-1, vmax=1)
fig.colorbar(cax)

columnas = matriz_corr.columns
ax.set_xticks(range(len(columnas)))
ax.set_yticks(range(len(columnas)))
ax.set_xticklabels(columnas, rotation=45, ha='left')
ax.set_yticklabels(columnas)

for i in range(len(columnas)):
    for j in range(len(columnas)):
        ax.text(j, i, f"{matriz_corr.iloc[i, j]:.2f}", 
                ha='center', va='center', 
                color='black' if abs(matriz_corr.iloc[i, j]) < 0.6 else 'white')

plt.title('Matriz de Correlación (matshow)', fontsize=14, fontweight='bold', pad=30)
plt.show()
#hola
```
